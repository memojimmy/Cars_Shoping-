```python
df['Power_bhp'] = df['Power_bhp'].str.replace('bhp','')
```

```python
df['Power_bhp'] = df['Power_bhp'].replace('null ',np.nan).astype('float64')
```

```python
df['Mileage'].unique()
```

```python
def clean_mileag (x):
    if type(x) == float :
        return np.nan
    if ' km/kg' in x :
        x = x.replace(' km/kg','')
        return float(x) * 1.35
    if  ' kmpl' in x :
        x = x.replace(' kmpl','')
        return float(x)
    return (x)
```

```python
df['Mileage'] =  df['Mileage'].apply(clean_mileag)
```

### numrical Pipeline

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline 

sc = StandardScaler()

num_pipeline = Pipeline(steps=[('Scaling' , sc)])
num_pipeline
```

## Categorical Pipeline
#### One Pipeline == 'Fuel_Type' ,'Transmission'
### BE Pipeline == 'Location' ,'Brand_name' , 'Model_name'
### ordinal Pipeline == 'Owner_Type' 

```python
nominal_col = x.select_dtypes('object').drop('Owner_Type',axis=1).columns
nominal_col
```

```python
from sklearn.preprocessing import OrdinalEncoder as SklearnOrdinalEncoder
from category_encoders import BinaryEncoder
from sklearn.pipeline import Pipeline
```

```python
for co in nominal_col: 
    print(co)
    print(df[co].nunique())
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline 

one = OneHotEncoder(drop='first' , sparse_output= False)

one_pipline = Pipeline(steps=[('Encoding one hot' , one)])
one_pipline
```

```python
from category_encoders import BinaryEncoder
from sklearn.pipeline import Pipeline 

bin = BinaryEncoder()

bin_pipeline = Pipeline(
    steps=[
        ('bin_encodeing', bin)
    ]
)
bin_pipeline
```

```python
x['Owner_Type'].unique()
```

```python
from sklearn.preprocessing import OrdinalEncoder as SklearnOrdinalEncoder
from sklearn.pipeline import Pipeline

sklearn_ord = SklearnOrdinalEncoder(
    categories=[
        ['Fourth & Above', 'Third', 'Second', 'First']
    ],
    handle_unknown='use_encoded_value',
    unknown_value=-1
)

ord_pipeline = Pipeline(
    steps=[
        ('ordinal_pip', sklearn_ord)
    ]
)

ord_pipeline
```

```python
print(num_col)
print(type(num_col))
```

```python
from sklearn.compose import ColumnTransformer 

preprocessing = ColumnTransformer([('numric', num_pipeline, num_col),
                                    ('one hot enc', one_pipline , ['Fuel_Type' ,'Transmission']),
                                    ('binary enc', bin_pipeline , ['Location' ,'Brand_name' , 'Model_name']),
                                    ('ordinal enc', ord_pipeline , ['Owner_Type'])
                                    ])
```
