```python
def clean_mileag (x):
    if type(x) == float :
        return np.nan
    if ' km/kg' in x :
        x = x.replace(' km/kg','')
        return float(x) * 1.35
    if  ' kmpl' in x :
        x = x.replace(' kmpl','')
        return float(x)
    return (x)
```

```python
df['Mileage'] =  df['Mileage'].apply(clean_mileag)
```

### Data preprocessing for ML with Pipeline

```python
x = df.drop('Price' ,axis=1)
y = df['Price']
```

```python
x.isna().sum()
```

```python
num_col = x.select_dtypes('number').columns
num_col
```

### numrical Pipeline

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline 

sc = StandardScaler()

num_pipeline = Pipeline(steps=[('Scaling' , sc)])
num_pipeline
```

```python
x.select_dtypes('object').isna().sum()
```

## Categorical Pipeline
#### One Pipeline == 'Fuel_Type' ,'Transmission'
### BE Pipeline == 'Location' ,'Brand_name' , 'Model_name'
### ordinal Pipeline == 'Owner_Type' 

```python
nominal_col = x.select_dtypes('object').drop('Owner_Type',axis=1).columns
nominal_col
```

```python
for co in nominal_col: 
    print(co)
    print(df[co].nunique())
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline 

one = OneHotEncoder(drop='first' , sparse_output= False)

one_pipline = Pipeline(steps=[('Encoding one hot' , one)])
one_pipline
```

```python
from category_encoders import BinaryEncoder
from sklearn.pipeline import Pipeline 

bin = BinaryEncoder()

bin_pipeline = Pipeline(steps=[('bin_encodeing' , bin)])

bin_pipeline
```

```python
x['Owner_Type'].unique()
```

```python
from sklearn.preprocessing import OrdinalEncoder 
from sklearn.pipeline import Pipeline 

ord = OrdinalEncoder(categories=[['Fourth & Above', 'Third', 'Second' ,'First']])

ord_pipeline = Pipeline(steps=[('ordinal_pip', ord)])

ord_pipeline
```

```python
print(num_col)
print(type(num_col))
```

```python
from sklearn.compose import ColumnTransformer 

preprocessing = ColumnTransformer([('numric', num_pipeline, num_col),
                                    ('one hot enc', one_pipline , ['Fuel_Type' ,'Transmission']),
                                    ('binary enc', bin_pipeline , ['Location' ,'Brand_name' , 'Model_name']),
                                    ('ordinal enc', ord_pipeline , ['Owner_Type'])
                                    ])
```

### Assign each columns to the corresponding Pipeline

```python
preprocessing
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
from sklearn.linear_model import LinearRegression ,  SGDRegressor , ElasticNet , Lasso , Ridge
from sklearn.neighbors import KNeighborsRegressor
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import cross_validate 
from sklearn.model_selection import KFold
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor 
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRFRegressor
from catboost import CatBoostRegressor
from lightgbm import LGBMRegressor 

models = [ ('linear', LinearRegression()) ,
        #   ('SGDRegressord' , SGDRegressor(random_state=42)) ,
          ('poly' ,  PolynomialFeatures(degree=2)),
        #   ('elastic' , ElasticNet(random_state=42))  , 
        #  ('lasso', Lasso(random_state=42)), 
          ('knn', KNeighborsRegressor(n_neighbors=5)),
          ('svm', SVR(kernel='linear')),
          ('DT', DecisionTreeRegressor(random_state=42)), 
          ('RF', RandomForestRegressor(random_state=42 , n_jobs= -1)),
          ('xgboost', XGBRFRegressor(random_state=42 , n_jobs= -1)),
          ('catboost', CatBoostRegressor(random_state=42)),
          ('lightgbm', LGBMRegressor(random_state=42 , n_jobs= -1)) 
          ]

for model in models : 

    model_pip = Pipeline(steps= [('preprocessing' , preprocessing) , ('models' , model[1])])

    if model[0] == 'poly':
        model_pip = Pipeline(steps=[('preprocessing' , preprocessing ), ('poly' , PolynomialFeatures()) , ('model' , Ridge())])

    cvs = cross_validate(estimator= model_pip , X=x , y= y , cv= 5 , scoring= 'r2' ,return_train_score=True)
    print(model[0])
    print( 'Train_score R2 ',round(cvs['train_score'].mean()*100, 2))
    print( 'Test_score R2 ',round(cvs['test_score'].mean()*100, 2))
    print( 'Fit_time  ',round(cvs['fit_time'].mean(), 4))
    print('\n', 100*'=' )
```

```python
from catboost import CatBoostRegressor
catboost_pipeline = Pipeline(steps=[('preprocessing' , preprocessing ), ('model' , CatBoostRegressor())])
catboost_pipeline
```

```python
from sklearn.model_selection import GridSearchCV 
param_grids = {
    "model__depth": [4, 5, 6, 7],
    "model__learning_rate": [0.01, 0.03, 0.05],
    "model__iterations": [1000, 1500, 2000], }

Grid_search = GridSearchCV(catboost_pipeline , param_grid=param_grids ,cv=5 , scoring='r2',return_train_score=True, n_jobs=-1 )
Grid_search.fit(x , y)
```

```python
Grid_search.best_params_
```

```python
Grid_search.best_score_
```

```python
catboost_finall_pipeline = Pipeline(steps=[('preprocessing' , preprocessing ), ('model' , CatBoostRegressor(random_state=42))])
catboost_finall_pipeline
```

```python
catboost_finall_pipeline.fit(x , y)
```
